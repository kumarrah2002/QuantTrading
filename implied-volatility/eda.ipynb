```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import os

pd.set_option('display.max_columns', None)
```

```python
folder_path = "data"
df = pd.DataFrame()

for filename in os.listdir(folder_path):
    file_path = os.path.join(folder_path, filename)
    if os.path.isfile(file_path) and filename.endswith('.par'):
        temp_df = pd.read_parquet(file_path)
        df = pd.concat([df, temp_df], ignore_index=True)
```

```python
base_features = ['symbol', 'RV21', 'RV63', 'VolRatio3_21', 'VolRatio5_63', 'DolVol_21', 'timestamp']

option_cols = {
    i: [f"{prefix}_{i}" for prefix in [
        'expiry', 'strike', 'type', 'implied_vol', 'delta', 'option_price',
        'pv_dividend', 'gamma', 'vega', 'theta', 'underlying_price',
        'day1_expiry', 'day1_strike', 'day1_type', 'day1_implied_vol', 'day1_delta', 'day1_option_price',
        'day1_pv_dividend', 'day1_gamma', 'day1_vega', 'day1_theta', 'day1_underlying_price'
    ]] for i in range(4)
}

df_options = {}

for i in range(4):
    cols = base_features + option_cols[i]
    df_options[i] = df[cols].copy()

df_opt0 = df_options[0]
df_opt1 = df_options[1]
df_opt2 = df_options[2]
df_opt3 = df_options[3]
```
